```python
import pandas as pd

df = pd.read_csv("../data/raw/Reviews.csv")

df = df.dropna()

```

```python
df = df[["ProductId", "UserId", "Score", "Summary", "Text"]]

user_counts = df["UserId"].value_counts()
product_counts = df["ProductId"].value_counts()

active_users = user_counts[user_counts >= 5].index
popular_products = product_counts[product_counts >= 5].index

df_filtered = df[
    df["UserId"].isin(active_users) &
    df["ProductId"].isin(popular_products)
]
```

```python
print(df_filtered.shape)
print(df_filtered["UserId"].nunique())
print(df_filtered["ProductId"].nunique())
```

```python
print(
    "Sparsity:",
    100 * (1 - len(df_filtered) /
    (df_filtered["UserId"].nunique() *
     df_filtered["ProductId"].nunique()))
)
```

```python
df_filtered.to_csv(
    "../data/processed/cleaned_reviews.csv",
    index=False
)
```
